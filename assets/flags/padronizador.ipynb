```python
from pathlib import Path
```

```python
ROOT_DIR = Path(
    r"C:\Users\corps\Documents\Cursor\github\organizador\assets\flags"
)

ROOT_DIR
```

```python
def normalizer(name: str) -> str:
    """
    Converter para lowcase e substituir espaços para underline
    """
    return name.lower().replace(" ", "_")
```

```python
def renamer(path: Path):
    new_name = normalizer(path.name)
    
    if path.name == new_name:
        return
    
    temp = path.with_name(f"_temp_{new_name}")
    
    path.rename(temp)
    temp.rename(path.with_name(new_name))
```

```python
paths = sorted(
    ROOT_DIR.rglob("*"),
    key=lambda p: len(p.parts),
    reverse=True
)

c = 0;

for path in paths:
    renamer(path)
    c+=1;
    
print(f"✅{c} arquivos renomeados com sucesso.")
```
